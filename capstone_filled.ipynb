## 5. Limitations

This work has several important limitations:

- The analysis is based on observational production/search data and therefore does not establish causal effects.
- Predictive performance depends on the selected validation design and may not generalize to every future dataset, client, or search environment.
- Features that were unavailable at prediction time or that could create leakage must be excluded, which limits the information available to the model.
- Model errors remain and may be concentrated in particular segments of the data.
- Associations identified by the model should be treated as hypotheses or signals for further testing, not guaranteed optimization levers.
- Public-safety requirements prevent publication of client names, private URLs, private queries, and other identifying information.
- Any recommendation should therefore be validated through controlled experiments or additional analysis before being treated as a causal intervention.

### Honest framing

The strongest defensible language is:

**observed, measured, predictive, associated with, directional, suggests, decision-support, worth testing.**

Avoid:

**proves, causes, guarantees, always, will increase, ranking factor** — unless independently supported by evidence that actually establishes those claims.
